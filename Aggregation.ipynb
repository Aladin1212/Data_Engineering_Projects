```python
from pyspark.sql.functions import *
from pyspark.sql.types import * 
```

```python
emp_data = [
(1,'manish',26,20000,'india','IT'),
(2,'rahul',None,40000,'germany','engineering'),
(3,'pawan',12,60000,'india','sales'),
(4,'roshini',44,None,'uk','engineering'),
(5,'raushan',35,70000,'india','sales'),
(6,None,29,200000,'uk','IT'),
(7,'adam',37,65000,'us','IT'),
(8,'chris',16,40000,'us','sales'),
(None,None,None,None,None,None),
(7,'adam',37,65000,'us','IT')
]

emp_schema = """
    Id INT, Name STRING, Age INT, Salary INT , Country STRING, Department STRING
"""
```

```python
emp_df = spark.createDataFrame(emp_data, emp_schema)
```

```python
emp_df.show(10)
```

```python
emp_df.printSchema()
```

```python
emp_df.count()  # It will Count total row including NUll
```

```python
emp_df.count("Name") # It will count only where name is exists 
```

```python
emp_df.count(col("Name")) # It will count only where name is exists 
```

```python
emp_df.("Name").count()  # It will count only where name is exists 
```

```python
emp_df.select("Name").count().show()  # It will count only where name is exists 
```

```python
 emp_df.select(count("Name")).show() # It will count only where name is exists 
```

```python
emp_df.select(count("Id")).show()
```

```python
emp_df.select(sum("Salary").alias("Sum_Salary"), MAX("Salary").alias("Max_Salary")).show()
```

```python
emp_df.select(sum("Salary").alias("Sum_Salary"), max("Salary").alias("Max_Salary"), min("salary").alias("Min_Salary")).show()
```

```python
emp_df.select(sum("salary").alias("Sum_Salary"), count("Salary"), avg("Salary").cast("int").alias("Avg_Salary")).show()
```
